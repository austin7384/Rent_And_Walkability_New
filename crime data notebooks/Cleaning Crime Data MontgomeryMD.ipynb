```python
import pandas as pd
import numpy as np
import geopandas as gpd
from shapely.geometry import Point
```

```python
dfCrime = pd.read_csv('/Users/austincoffelt/Documents/Rent-Walk-Local/Crime.csv')
print(dfCrime.shape)
dfCrime.head()
```

```python
dfCrime['Start_Date_Time'] = pd.to_datetime(dfCrime['Start_Date_Time'])
dfCrime.head()
```

```python
dfCrime = dfCrime.loc[(dfCrime['Start_Date_Time'] >= '2017-01-01') & (dfCrime['Start_Date_Time'] < '2020-01-01')]
print(dfCrime.shape)
dfCrime.head()
```

```python
list(dfCrime['Crime Name2'].unique())
```

```python
violentCrimes = ['Robbery', 'Aggravated Assault', 'Forcible Rape', 'Murder and Nonnegligent Manslaughter', 'Human Trafficking, Commercial Sex Acts']
propertyCrimes = ['Burglary/Breaking and Entering', 'Theft From Motor Vehicle', 'Theft from Building', 'All other Larceny', 
                  'Stolen Property Offenses', 'Theft of Motor Vehicle Parts or Accessories', 'Motor Vehicle Theft']
both = violentCrimes + propertyCrimes
dfCrime = dfCrime[dfCrime['Crime Name2'].isin(both)]
dfCrime = dfCrime[['Crime Name2', 'Longitude', 'Latitude']]
print(dfCrime.shape)
dfCrime.head()
```

```python
# make our lon and lat into a point type for spacial join
dfCrime['geometry'] = dfCrime.apply(lambda row: Point(row['Longitude'], row['Latitude']), axis=1)
dfCrime = dfCrime[['Crime Name2', 'geometry']]
print(dfCrime.shape)
dfCrime.head()
```

```python
# Read shapefile (automatically loads .shp, .shx, .dbf)
gdf = gpd.read_file("/Users/austincoffelt/Documents/Rent-Walk-Local/cb_2018_24_bg_500k/cb_2018_24_bg_500k.shp")

# View the first 5 rows
print(gdf.shape)
gdf.head()
```

```python
# make sure points match the crs of gdf
gdfPoints = gpd.GeoDataFrame(
    dfCrime, 
    geometry='geometry',
    crs="EPSG:4269"  # WGS84 coordinate reference system
)
gdfPoints.head()
```

```python
# Basic spatial join (points within polygons)
result = gpd.sjoin(
    gdfPoints, 
    gdf,
    how="left",       # 'left', 'right', or 'inner'
    predicate="within" # Spatial relationship
)
print(result.shape)
result.head()
```

```python
crime = result['AFFGEOID'].value_counts() / 3
crime
```

```python
crime.to_csv('/Users/austincoffelt/Documents/RW Data/MontMD_all_crime_by_BG.csv')
```
